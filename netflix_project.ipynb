```python
import numpy as np
import pandas as pd
```

```python
# load dataset
netflix = pd.read_csv("netflix_titles.csv")
```

```python
# display dataset
netflix.head()
```

```python
# check dataset information
netflix.shape
```

```python
netflix.info()
```

```python
# check missing values
netflix.isnull().sum()
```

```python
# remove null values
# Question 1: Remove null values from the Netflix dataset.

df_clean = netflix.dropna()

df_clean
```

```python
# verify null values are removed

df_clean.isnull().sum()
```

```python
# Find Movies vs TV Shows in the USA
us_content = df_clean.query(
    "country.str.contains('United States')",
    engine='python'
)

us_content
```

```python
# Count the number of Movies and TV Shows in the USA.

us_content['type'].value_counts()
```

```python
# Find the Average Rating of Movies
movies = df_clean.query("type == 'Movie'")

movies
```

```python
# check the rating
movies['rating'].value_counts()
```

# project conclusion

 Netflix provides categorical maturity ratings rather than numerical ratings. Therefore, calculating the average movie rating using mean() is not possible. We can use value_counts() to understand the distribution of movie ratings
